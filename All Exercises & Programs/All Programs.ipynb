# 2. Tuple Data Type, Packing, Unpacking & Methods

```python
# Tuple
# Important Point: Parentheses () are commonly used to create tuples
data = (10, 20, 30, 40, 50)
print(type(data))
```

```python
# But parentheses are not actually mandatory
# The comma , is what makes a tuple, not the parentheses.
data2 = 10, 20, 30, 40, 50
print(type(data2))
```

```python
data3 = (10)
print(type(data3))
# Output: <class 'int'>

# Why?
# Because (10) is not a tuple. It is simply an integer 10 enclosed in parentheses.

# Python treats it as:
# data3 = 10
```

```python
# How to create a single-element tuple?
# You must use a comma:
data4 = (10,)
print(type(data4))

# Remember this rule:
# Code          Type
# data = (10)   int
# data = (10,)  tuple
# data = 10,    tuple
# data = (10, 20) tuple
```

```python
# Tuple Indexing
# Tuples are ordered and immutable.
data = (10, 20, 30, 40, 50)
print(data[2]) # Indexing is used to access individual elements of a tuple and Tuple indexing starts from 0.
print(data[-1]) # Negative indexing starts from -1 from the right side.
```

```python
# A tuple allows duplicate elements.
data = (10, 20, 30, 40, 50, 30) # Duplicate Elements
print(data)
```

```python
# Tuple is Immutable
# A tuple is immutable, which means its elements cannot be changed, updated, or deleted after creation.
data = (10, 20, 30, 40, 50)
data[0] = 88 # tries to change the first element from 10 to 88. Therefore, Python raises a TypeError.

# TypeError: 'tuple' object does not support item assignment
```

```python
# Tuple Packing:
studentData = "Aman", "MCA AI", 78.65 # Tuple Packing: Tuple Packing means storing multiple values into a single tuple without explicitly using parentheses.
print(type(studentData))

# Tuple Unpacking
name, course, avgMarks = studentData # Tuple Unpacking means extracting the elements of a tuple into separate variables.
print(name)
# Note: The number of variables should normally match the number of tuple elements.
```

```python
record = "Amit", "MCA", 23, True # Tuple Packing
print(record[0])
print(record[1])
print(record)

name, course, age, att = record
# name, course, age, att = "Amit", "MCA", 23, True    Tuple Unpacking

print(name)

# Unpacking Using *
# * in tuple unpacking is called extended unpacking. It allows one variable to collect multiple remaining values.
n, *ab = record
# Note: The * operator collects the remaining elements into a list.
print(n)
print(ab)
```

```python
record = "Amit", "MCA", 23, True # Tuple Packing
name, course, age = record # ValueError: too many values to unpack (expected 3)
# Note: The number of variables should normally match the number of tuple elements.
```

```python
record = "Amit", "MCA", 23, True # Tuple Packing
name, *ab, att = record
print(name)
print(ab)
print(att)
```

```python
# len() Returns the total number of elements
data = (10, 20, 30, 40, 50, 30)
print(len(data))

# count() : Counts how many times a value occurs
print(data.count(30))

print(sum(data)) # Returns the sum of numeric elements
print(max(data)) # Returns the largest element
print(min(data)) # Returns the smallest element
print(data.index(10)) # Returns the index of the first occurrence
```

```python
# Arithmetic Operators
x, y = 10, 20
print(x + y)
print(x - y)
print(x * y)
print(x / y)

# floor division
print(7 // 2) # 3    Floor division returns the floor value (greatest integer less than or equal to the result).
print(-7 // 2) # -4
print(2 ** 3) # 8    Exponentiation ** is used to calculate the power of a number.

# Important Note: Floor division does not simply remove the decimal part; it moves the result toward negative infinity.
```

```python
# String Concatenation with +
fistName = "Amit"
lastName = "Kumar"
print(fistName + lastName)
```

```python
# Compound Assignment Operators: Compound assignment operators are used to perform an operation and assign the result back to the same variable.

x, y = 40, 50
x += y # x = x + y >>> x = 40 + 50 >>> x = 90
print(x)

x -= y # x = x - y >>> x = 90 - 50 >> x = 40
print(x)

x *= y # x = x * y >> x = 40 * 50 >>> x = 2000
print(x)

x /= y # x = x / y >> x = 2000 / 50 >> 40.0
print(x)
```

```python
# Relational / Comparison Operators in Python. Relational operators are used to compare two values.
# The result of every relational operator is always a Boolean value: True or False
x = 10
y = 20

print(x == y) # Equal to == operator checks whether two values are equal.
print(x > y) # Greater Than > operator checks whether the left-hand value is greater than the right-hand value.
print(x < y) # Less Than < operator checks whether the left-hand value is less than the right-hand value.
print(x >= y) # Greater Than or Equal To >= operator checks two conditions:
              # Left value is greater than the right value
              # OR both values are equal
print(x <= y) # Less Than or Equal To <= operator checks two conditions:
              # Left value is less than the right value
              # OR both values are equal
print(x != y) # Not Equal To != operator checks whether two values are different.
```

```python
# Truth Tables for Logical AND and OR
# AND
# FC      SC      Result
# T       T       T
# T       F       F
# F       T       F
# F       F       F

# OR
# T       T       T
# T       F       T
# F       T       T
# F       F       F
```

```python
# Logical and Operator in Python:
# The and operator is used to combine two or more conditions. and = BOTH conditions must be True.
x = 10
y = 20
print(x < y and y > x)
print(x < y and x == y)
print(x > y and x < y)
print(x > y and x == y)
```

```python
# Identity Operator (is) and Equality Operator (==)
x = [10, 20] # Python creates a list object in memory.
y = x # Python does not create another list. Instead, y refers to the same list object.
# So, both x and y point to the same object.
print(x is y) # True Why? Because x and y refer to the same list object.
# is Operator: The is operator checks whether two variables refer to the same object in memory.
print(id(x))
print(id(y))
print(x == y) # == Operator checks value/content equality.
```

```python
# == vs is with List
x = [10, 20] # Python creates a list object in memory and x refers to that object.
y = [10, 20] # Python creates another separate list object. Although the values are the same, the two lists are different objects.
             # Because list belongs to mutable Type.
print(x == y) # Output: True   Why? The == operator compares the values/content of two objects.
print(x is y) # Output: False  Why? The is operator checks object identity.
x.append(33) # The list referred to by x is modified. Only the x list changes because x and y refer to different list objects.
print(x == y) # Output: False  Because: x -> [10, 20, 33]   y -> [10, 20] Their contents are now different.
print(id(x))
print(id(y))
```

```python
x = 10 # x refers to the integer object 10.
y = 10 # Python can make y refer to the same integer object 10.
print(id(x))
print(id(y))
print(x == y) # Therefore: id(x) == id(y) is typically: True
print(x is y) # And: x is y returns: True because x and y refer to the same object.

# Note: For integers and some other immutable objects, Python may reuse objects, but you should generally use == to compare values
# and use is mainly when checking object identity
```

## Conditional Exercises — Set 1

```python
# Exercise 1: ATM Withdrawal
# Write a Python program for ATM withdrawal.
# Conditions:
# Check whether PIN is correct.
# If PIN is correct:
#   Check whether withdrawal amount is a multiple of 100.
#   If yes, check whether balance is sufficient.
#   If sufficient, withdraw the amount.
#   Otherwise print "Insufficient Balance".
# If PIN is incorrect, print "Invalid PIN".

balance = 25000
pin = int(input("Enter Your Pin: "))
if(pin == 1234):
    amount = int(input("What amount do you want to Withdraw: "))
    if(amount % 100 == 0):
        if(amount <= balance):
            balance -= amount # balance = balance - amount
            print("Withdraw amount is: ", amount)
            print("Remaining balance is: ", balance)
        else:
            print("Insufficient Balance")
    else:
        print("Invalid amount")
else:
    print("Invalid PIN")
```

```python
# Exercise 2: Student Result System
# Write a program that takes marks of 3 subjects.
# Conditions:
# If every subject has marks >= 40:
#   Calculate percentage.
#   If percentage >= 75 -> Distinction (A+)
#   If percentage >= 60 -> First Division
#   If percentage >= 50 -> Second Division
#   Otherwise -> Pass
# If any subject is below 40 -> Fail.

java = int(input("Enter Marks of Java: "))
python = int(input("Enter Marks of Python: "))
dbms = int(input("Enter Marks of DBMS: "))

if(java >= 40):
    if(python >= 40):
        if(dbms >= 40):
            totalMarks = java + python + dbms
            per = totalMarks / 3
            print("Total is:", totalMarks)
            print("Student Per :", per)
            if(per >= 75):
                print("A+")
            elif(per >= 60):
                print("First Division")
            elif(per >= 50):
                print("Second Division")
            else:
                print("Pass")
        else:
            print("Fail in DBMS")
    else:
        print("Fail in Python")
else:
    print("Fail in Java")
```

```python
# Exercise 3: Bank Loan Eligibility
# A bank gives a loan based on:
# Age must be >= 21.
# If age is valid:
#   Salary must be >= 25,000.
#   If salary is valid:
#     Credit score >= 750 -> Loan approved (Interest Rate: 8%)
#     Credit score >= 650 -> Loan approved with higher interest (Interest Rate: 10%)
#     Otherwise -> Loan rejected.

age = int(input("Enter Your Age: "))
salary = int(input("Enter Your Salary: "))
creditScore = int(input("Enter Your Credit Score: "))

if(age >= 21):
    if(salary >= 25000):
        if(creditScore >= 750):
            print("Loan Approved")
            print("Interest Rate: 8%")
        elif(creditScore >= 650):
            print("Loan Approved")
            print("Interest Rate: 10%")
        else:
            print("Loan Rejected")
    else:
        print("Salary is too Low")
else:
    print("Age is not eligible")
```

```python
# Exercise 4: E-Commerce Discount
# An online shopping website provides discount based on:
# If cart amount >= 5000:
#   If customer is a premium member:
#     Discount = 20%
#   Otherwise:
#     Discount = 10%
# If cart amount < 5000:
#   If premium member:
#     Discount = 5%
#   Otherwise:
#     No discount.
# Print final amount.

amount = 8000
pMember = True
if(amount >= 5000):
    if(pMember):
        discount = amount * 20 / 100
        print("Discount amount is: ", discount)
    else:
        discount = amount * 10 / 100
        print("Discount amount is: ", discount)
elif(pMember):
    discount = amount * 5 / 100
    print("Discount amount is: ", discount)
else:
    discount = 0
    print("Discount amount is: ", discount)

final_amount = amount - discount
print("Your Final Amount is: ", final_amount)
```

```python
# Exercise 5: Login + Role Based Access
# Create a login system.
# Conditions:
# Check username.
# If username is correct:
#   Check password.
#   If password is correct:
#     Check role:
#       "Admin" -> You Have Full Access
#       "Trainer" / "Teacher" -> Trainer Dashboard
#       "Student" -> Student Dashboard
#       Anything else -> Invalid Role
#   Wrong password -> "Invalid Password"
# Wrong username -> "Invalid User Name"

userName = input("Enter UserName: ")
password = input("Enter Password: ")
role = input("Enter Role: ")
if(userName == "Admin"):
    if(password == "12345"):
        if(role == "Admin"):
            print("You Have Full Access")
        elif(role == "Trainer"):
            print("Trainer Dashboard")
        elif(role == "Student"):
            print("Student Dashboard")
        else:
            print("Invalid Role")
    else:
        print("Invalid Password")
else:
    print("Invalid User Name")
```

```python
# Exercise 6: Movie Ticket Booking
# Create a ticket booking system.
# Conditions:
# Check whether seats are available.
# If available:
#   Check customer age.
#   If age < 5 -> Free
#   If age 5-12 -> 50% discount
#   If age > 60 -> 30% discount
#   Otherwise -> Full price.
# If no seats -> "House Full".

seatsAvailable = True
ticketPrice = 300
age = int(input("Enter Customer Age: "))

if seatsAvailable:
    if age < 5:
        print("Ticket is Free! Final Price: 0")
    elif age <= 12:
        finalPrice = ticketPrice * 0.50
        print("50% Discount Applied! Final Price:", finalPrice)
    elif age > 60:
        finalPrice = ticketPrice * 0.70
        print("30% Discount Applied! Final Price:", finalPrice)
    else:
        print("Full Price:", ticketPrice)
else:
    print("House Full")
```

## Conditional Exercises — Set 2

```python
# Exercise 1: Electricity Bill
# Conditions:
# 0-100     -> Rs. 5/unit
# 101-200   -> Rs. 7/unit
# 201-300   -> Rs. 10/unit
# Above 300 -> Rs. 15/unit
# Additionally:
# Bill > Rs. 3000 -> 10% surcharge

units = int(input("Enter Electricity Units Consumed: "))

if units <= 100:
    bill = units * 5
elif units <= 200:
    bill = (100 * 5) + ((units - 100) * 7)
elif units <= 300:
    bill = (100 * 5) + (100 * 7) + ((units - 200) * 10)
else:
    bill = (100 * 5) + (100 * 7) + (100 * 10) + ((units - 300) * 15)

if bill > 3000:
    surcharge = bill * 0.10
    bill += surcharge
    print("Surcharge (10%):", surcharge)

print("Total Electricity Bill:", bill)
```

```python
# Exercise 2: Employee Bonus
# Conditions:
# Experience >= 1 year:
#   Salary >= 50,000:
#     Performance >= 90 -> 20%
#     Performance >= 75 -> 10%
#     Otherwise -> 5%
#   Salary < 50,000:
#     Performance >= 90 -> 15%
#     Otherwise -> 5%
# Experience < 1 year -> No Bonus

experience = int(input("Enter Experience (in years): "))
salary = int(input("Enter Salary: "))
performance = int(input("Enter Performance Score: "))

if experience >= 1:
    if salary >= 50000:
        if performance >= 90:
            bonus = salary * 0.20
        elif performance >= 75:
            bonus = salary * 0.10
        else:
            bonus = salary * 0.05
    else:
        if performance >= 90:
            bonus = salary * 0.15
        else:
            bonus = salary * 0.05
    print("Bonus Amount:", bonus)
else:
    print("Not Eligible for Bonus (Experience < 1 year)")
```

```python
# Exercise 4: ATM - Difficult Challenge
# Conditions:
# PIN correct
# Account active
# Amount positive
# Amount multiple of 100
# Balance sufficient
# Minimum Rs. 500 balance must remain after withdrawal
# Amount > 20,000 -> Extra verification

pin = int(input("Enter PIN: "))
accountActive = True
balance = 30000

if pin == 1234:
    if accountActive:
        amount = int(input("Enter Withdrawal Amount: "))
        if amount > 0:
            if amount % 100 == 0:
                if amount <= balance and (balance - amount) >= 500:
                    if amount > 20000:
                        otp = int(input("Amount > 20000. Enter OTP (1111) for Extra Verification: "))
                        if otp == 1111:
                            balance -= amount
                            print("Withdrawal Successful! Remaining Balance:", balance)
                        else:
                            print("Verification Failed! Invalid OTP")
                    else:
                        balance -= amount
                        print("Withdrawal Successful! Remaining Balance:", balance)
                else:
                    print("Insufficient Balance (Minimum Rs. 500 must remain)")
            else:
                print("Invalid Amount (Must be a multiple of 100)")
        else:
            print("Amount must be positive")
    else:
        print("Account is Inactive")
else:
    print("Invalid PIN")
```

## Dictionary Exercises (1 to 10)

```python
# 1. Check Student Result
# Question:
# Given a dictionary containing student marks, use a loop to print whether each student has Passed or Failed.
# Condition:
# Marks >= 40 -> Pass
# Marks < 40 -> Fail

students = {
    "Rahul": 75,
    "Amit": 32,
    "Priya": 88,
    "Neha": 39,
    "Rohit": 55
}

for name, marks in students.items():
    if marks >= 40:
        print(name, " Pass")
    else:
        print(name, " Fail")
```

```python
# 2. Student Performance Using Nested If
# Question:
# Print the performance category of each student.
# Conditions:
# Marks >= 80 -> Excellent
# Marks >= 60 -> Good
# Marks >= 40 -> Average
# Marks < 40 -> Fail

students = {
    "Rahul": 85,
    "Amit": 35,
    "Priya": 72,
    "Neha": 48
}

for name, Marks in students.items():
    if(Marks > 40):
        if(Marks >= 80):
            print(name, " Excellent")
        elif(Marks >= 60):
            print(name, " Good")
        elif(Marks >= 40):
            print(name, " Average")
    else:
        print(name, " Fail")
```

```python
# 3. Product Stock Management
# Question:
# A dictionary contains products and their available stock. Use a loop and nested conditions to display the stock status.
# Conditions:
# Stock = 0 -> Out of Stock
# Stock <= 5 -> Low Stock
# Stock > 5 -> Available

products = {
    "Laptop": 10,
    "Mouse": 3,
    "Keyboard": 0,
    "Monitor": 7,
    "Printer": 2
}

for product, stock in products.items():
    if stock > 0:
        if stock > 5:
            print(product, ": Available")
        else:
            print(product, ": Low Stock")
    else:
        print(product, ": Out of Stock")
```

```python
# 4. Employee Salary and Department
# Question:
# The following dictionary contains employee salary and department. Display the employee category.
# Conditions:
# Salary >= 50,000:
#   IT -> Senior IT Employee
#   HR -> Senior HR Employee
#   Other department -> Senior Employee
# Salary < 50,000 -> Junior Employee

employees = {
    "Rahul": {"salary": 60000, "department": "IT"},
    "Amit": {"salary": 40000, "department": "HR"},
    "Priya": {"salary": 70000, "department": "IT"},
    "Neha": {"salary": 45000, "department": "Finance"}
}

for name, data in employees.items():
    salary = data["salary"] # salary = 60000
    department = data["department"]

    if salary >= 50000:
        if department == "IT":
            print(name, " Senior IT Employee")
        elif department == "HR":
            print(name, " Senior HR Employee")
        else:
            print(name, " Senior Employee")
    else:
        print(name, " Junior Employee")
```

```python
# 5. Check Even/Odd Values
# Question:
# Given a dictionary, check whether each value is even or odd.
# If the value is:
# Even and greater than 20 -> Large Even
# Even and <= 20 -> Even
# Odd -> Odd

data = {
    "a": 10,
    "b": 15,
    "c": 25,
    "d": 30,
    "e": 40
}

for key, val in data.items():
    if val % 2 == 0:
        if val > 20:
            print(key, ":", val, "-> Large Even")
        else:
            print(key, ":", val, "-> Even")
    else:
        print(key, ":", val, "-> Odd")
```

```python
# 6. Shopping Cart Classification
# Question:
# A dictionary contains products and their prices.
# Conditions:
# Price >= 1000:
#   Price >= 5000 -> Premium Product
#   Otherwise -> Expensive Product
# Price < 1000 -> Affordable Product

products = {
    "Mobile": 15000,
    "Mouse": 500,
    "Laptop": 60000,
    "Keyboard": 1200
}

for item, price in products.items():
    if price >= 1000:
        if price >= 5000:
            print(item, ": Premium Product")
        else:
            print(item, ": Expensive Product")
    else:
        print(item, ": Affordable Product")
```

```python
# 7. Student Eligibility
# Question:
# The dictionary contains student attendance and marks.
# Conditions:
# Attendance >= 75:
#   Marks >= 40 -> Eligible
#   Marks < 40 -> Failed in Exam
# Attendance < 75 -> Not Eligible

students = {
    "Rahul": {"attendance": 85, "marks": 70},
    "Amit": {"attendance": 60, "marks": 80},
    "Priya": {"attendance": 90, "marks": 35},
    "Neha": {"attendance": 78, "marks": 55}
}

for name, info in students.items():
    if info["attendance"] >= 75:
        if info["marks"] >= 40:
            print(name, ": Eligible")
        else:
            print(name, ": Failed in Exam")
    else:
        print(name, ": Not Eligible")
```

```python
# 8. Bank Account Classification
# Question:
# A dictionary contains customer names and account balances.
# Conditions:
# Balance > 0:
#   Balance >= 10,000 -> Premium Account
#   Otherwise -> Regular Account
# Balance = 0 -> No Balance
# Balance < 0 -> Overdraft

accounts = {
    "Rahul": 25000,
    "Amit": 5000,
    "Priya": 0,
    "Neha": -2000
}

for customer, balance in accounts.items():
    if balance > 0:
        if balance >= 10000:
            print(customer, ": Premium Account")
        else:
            print(customer, ": Regular Account")
    elif balance == 0:
        print(customer, ": No Balance")
    else:
        print(customer, ": Overdraft")
```

```python
# 9. Nested Dictionary - Employee Experience
# Question:
# Classify employees based on salary and experience.
# Conditions:
# Salary >= 50,000:
#   Experience >= 5 -> Senior
#   Otherwise -> Mid-Level
# Salary < 50,000:
#   Experience >= 3 -> Experienced Junior
#   Otherwise -> Fresher/Junior

employees = {
    "Rahul": {"salary": 65000, "experience": 6},
    "Amit": {"salary": 55000, "experience": 3},
    "Priya": {"salary": 40000, "experience": 4},
    "Neha": {"salary": 30000, "experience": 1}
}

for name, info in employees.items():
    if info["salary"] >= 50000:
        if info["experience"] >= 5:
            print(name, ": Senior")
        else:
            print(name, ": Mid-Level")
    else:
        if info["experience"] >= 3:
            print(name, ": Experienced Junior")
        else:
            print(name, ": Fresher/Junior")
```

```python
# Question 10:
# The following dictionary contains students' marks and attendance.
# Conditions:
# If attendance < 75 -> Not Eligible
# If attendance >= 75:
#   Marks >= 80 -> Distinction
#   Marks >= 60 -> First Division
#   Marks >= 40 -> Pass
#   Marks < 40 -> Fail

students = {
    "Rahul": {"marks": 85, "attendance": 90},
    "Amit": {"marks": 35, "attendance": 80},
    "Priya": {"marks": 65, "attendance": 70},
    "Neha": {"marks": 45, "attendance": 85},
    "Rohit": {"marks": 90, "attendance": 60}
}

for name, info in students.items():
    if info["attendance"] < 75:
        print(name, "Not Eligible")
    else:
        if info["marks"] >= 80:
            print(name, "Distinction")
        elif info["marks"] >= 60:
            print(name, "First Division")
        elif info["marks"] >= 40:
            print(name, "Pass")
        else:
            print(name, "Fail")
```

# 9. Pattern Printing (`for` & `while` Loops)

```python
# Square Star Pattern (for loop)
n = 4
for i in range(1, n + 1):
    for j in range(1, n + 1):
        print("*", end=" ")
    print(end="\n")
```

```python
# Rectangle Star Pattern (for loop)
row = 4
column = 6
for i in range(1, row + 1):
    for j in range(1, column + 1):
        print("*", end=" ")
    print(end="\n")
```

```python
# Right-Angled Triangle Star Pattern (for loop)
n = 5
for i in range(1, n + 1):
    for j in range(1, i + 1):
        print("*", end=" ")
    print(end="\n")
```

```python
# Inverted Right-Angled Triangle Star Pattern (for loop)
n = 5
for i in range(n, 0, -1):
    for j in range(1, i + 1):
        print("*", end=" ")
    print(end="\n")
```

```python
# Number Triangle Pattern - Column-wise 1 to i (for loop)
n = 5
for i in range(1, n + 1):
    for j in range(1, i + 1):
        print(j, end=" ")
    print(end="\n")
```

```python
# Number Triangle Pattern - Row-wise repeating i (for loop)
n = 5
for i in range(1, n + 1):
    for j in range(1, i + 1):
        print(i, end=" ")
    print(end="\n")
```

```python
# Floyd's Triangle Pattern (1 to 15)
n = 5
x = 1
for i in range(1, n + 1): # i = 3
    for j in range(1, i + 1):
        print(x, end=" ")
        x += 1
    print(end="\n")
```

```python
# Alternating 0-1 Triangle Pattern
n = 5
for i in range(1, n + 1):
    for j in range(0, i):
        print((i + j) % 2, end=" ")
    print(end="\n")
```

```python
# Alphabet Triangle Pattern (A to E)
n = 5
for i in range(1, n + 1): # i = 2
    for j in range(0, i): # j = 0, j = 1 , i = 2   Loop 2 time
        print(chr(65 + j), end=" ") # 65 + 1
    print(end="\n")
```

```python
# Diamond Star Pattern (Mirror Pattern)
n = 5
for i in range(1, 2 * n): # i = 7 row = 7 to row 9
    if i <= n: # 7 <= 5
        star = i # 1st row no of star 5
    else:
        star = 2 * n - i # 7th no row star = (2*5) - 7 = 3 star
    space = n - star # space = 5 - 3        space = 2      7th row

    for j in range(0, space): # 2 time
        print("  ", end="")

    for k in range(0, 2 * star - 1): # (2*3) - 1 = 5 >> 5 time
        print("* ", end="")

    print("\n")
```

```python
# Mirror Number Triangle Pattern (1 to 5 and back to 1)
n = 5
for i in range(1, 2 * n): # i = 9   row 9
    if i <= n:
        num = i
    else:
        num = 2 * n - i # num = (2*5) - 9      num = 1
    for j in range(1, num + 1):
        print(j, end=" ")

    print(end="\n")
```

```python
# Reverse Mirror Number Triangle Pattern (num down to 1)
n = 5
for i in range(1, 2 * n):
    if i <= n:
        num = i
    else:
        num = 2 * n - i
    for j in range(num, 0, -1):
        print(j, end=" ")
    print(end="\n")
```

```python
# Square Star Pattern using while loop
n = 4
row = 1

while(row <= n):
    star = 1
    while(star <= n):
        print("* ", end=" ")
        star += 1
    print(end="\n")
    row += 1
```

```python
# Right-Angled Triangle Star Pattern using while loop
num = 5
row = 1

while(row <= num):
    star = 1
    while(star <= row):
        print("* ", end=" ")
        star += 1
    print(end="\n")
    row += 1
```

```python
# Inverted Right-Angled Triangle Star Pattern using while loop
num = 5
row = 1
while(row <= num):
    star = 1
    while(star <= num - row + 1):
        print("* ", end=" ")
        star += 1
    print(end="\n")
    row += 1
```

```python
# Right-Aligned Triangle Star Pattern using while loop
num = 5
row = 1
while(row <= num):
    space = 1
    while(space <= num - row):
        print(" ", end=" ")
        space += 1
    star = 1
    while(star <= row):
        print("*", end=" ")
        star += 1
    print(end="\n")
    row += 1
```

```python
# Inverted Right-Aligned Triangle Star Pattern using while loop
num = 5
row = 1
while(row <= num):
    space = 1
    while(space <= row - 1):
        print(" ", end=" ")
        space += 1
    star = 1
    while(star <= num - row + 1): # 5 - 1 + 1 = 5
        print("*", end=" ")
        star += 1
    print(end="\n")
    row += 1
```

```python
# Pyramid Star Pattern using while loop
num = 5
row = 1
while(row <= num):
    space = 1
    while(space <= num - row): # 5 - 1 = 4
        print(" ", end=" ")
        space += 1
    star = 1
    while(star <= 2 * row - 1): # (2*1) - 1 = 1
        print("*", end=" ")
        star += 1
    print(end="\n")
    row += 1
```

# 10. List Comprehension & Exercises

```python
# Filtering list elements using a standard for loop
data = [10, 20, 30, 40, 50]
for i in data:
    if i >= 30:
        print(i)
```

```python
# Squaring elements using a standard for loop
data = [1, 2, 3, 4, 5]
empty = []
for i in data:
    x = i ** 2
    empty.append(x)
print(empty)
```

```python
# Squaring elements using List Comprehension
data = [1, 2, 3, 4, 5]
empty = [x ** 2 for x in data]
print(empty)
```

```python
# Filtering even numbers using List Comprehension
data = [23, 17, 44, 66, 21, 8]
even = [x for x in data if x % 2 == 0]
print(even)
```

```python
# Pass/Fail classification using List Comprehension (if-else)
data = [76, 45, 80, 34, 23, 85]
result = ["Pass" if x >= 60 else "Fail" for x in data]
print(result)
```

```python
# Converting strings to uppercase using List Comprehension
names = ["Amit", "Priya", "Rahul", "Soniya", "Sumit", "John"]
result = [n.upper() for n in names]
print(result)
```

```python
# Cleaning whitespace from strings using List Comprehension
names = [" Amit ", "   Priya ", " Rahul", "  Soniya ", "Sumit  ", "John"]
cleand_Data = [x.strip() for x in names]
print(names)
print(cleand_Data)
```

```python
# Extracting values from a list of dictionaries using List Comprehension
users = [
    {"name": "Rahul", "age": 25},
    {"name": "Priya", "age": 22}
]

# [Rahul, Priya]
empNames = [x["name"] for x in users]
print(empNames)
```
