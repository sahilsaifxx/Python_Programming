```python
# 1. Basic QR Code Generator (Contact / Email / Phone)
import qrcode

data = input("Enter Number/Email/Text to Make QR: ")
qr = qrcode.make(data) # qrcode

qr.save("Sahil_Contact_QR.png") # pillow

print("QR Code Generated Successfully!")
display(qr)
```

```python
# 2. Generate QR Code for a Valid URL (http / https)
import qrcode

url = input("Enter URL to Make QR: ")

if url.startswith("http://") or url.startswith("https://"):
    qr = qrcode.make(url) # qrcode
    qr.save("Sahil_Website_QR.png") # pillow
    display(qr)
    print("QR Code Generated Successfully!")
else:
    print("Invalid URL..!!! Must start with http or https")
```

```python
# 3. Customized QR Code (Version, Error Correction, Box Size, Border & Colors)
import qrcode

url = input("Enter URL to Make QR: ")

if url.startswith("http://") or url.startswith("https://"):
    qr = qrcode.QRCode(
        version=1,
        error_correction=qrcode.constants.ERROR_CORRECT_H,
        box_size=10, # 10x10
        border=4
    )
    qr.add_data(url)
    qr.make(fit=True)

    image = qr.make_image(fill_color="yellow", back_color="white")
    image.save("Sahil_Custom_QR.png")
    display(image)
    print("Custom Colored QR Code Generated Successfully!")
else:
    print("Invalid URL..!!! Must start with http or https")
```

```python
# 4. Create QR Code For UPI Payment
import qrcode

upi_id = input("Enter UPI ID: ")
name = input("Enter Name: ")

upi_url = f"upi://pay?pa={upi_id}&pn={name}"

qr = qrcode.make(upi_url)
qr.save("Sahil_UPI.png")

print("UPI QR Code Generated Successfully..!!")
display(qr)
```

```python
# 5. Create QR Code For UPI with Specific Amount
import qrcode

upi_id = input("Enter UPI ID: ")
name = input("Enter Name: ")
amount = int(input("Enter Amount: "))

upi_url = f"upi://pay?pa={upi_id}&pn={name}&am={amount}"

# pa = Payee Address (UPI ID)
# pn = Payee Name
# am = Amount
# cu = Currency (INR)

qr = qrcode.make(upi_url)
qr.save("Sahil_UPI_2.png")

print("UPI QR Code with Amount Generated Successfully..!!")
display(qr)
```
